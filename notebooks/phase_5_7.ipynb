```python
%%writefile src/business_entity_resolution/blocking.py

import re
import pandas as pd


def make_block_key(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = re.sub(r"[^a-z0-9]+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    if not text:
        return ""
    words = text.split()

    return " ".join(words[:2])


def create_blocking_keys(df):
    result = df.copy()

    result["name_block_key"] = (
        result["business_name"]
        .fillna("")
        .apply(make_block_key)
    )

    result["address_block_key"] = (
        result["business_address"]
        .fillna("")
        .apply(make_block_key)
    )

    result["country_block_key"] = (
        result["country"]
        .fillna("")
        .astype(str)
        .str.lower()
        .str.strip()
    )

    return result
```

```python
%%writefile src/business_entity_resolution/candidate_generation.py

import pandas as pd


def generate_candidates(source1, source2):

    s1 = source1.copy()
    s2 = source2.copy()

    candidate_pairs = set()
    name_index = {}

    for _, row in s2.iterrows():

        key = (
            row["country_block_key"],
            row["name_block_key"]
        )

        if key != ("", ""):
            name_index.setdefault(key, []).append(
                row["entity_id"]
            )

    address_index = {}

    for _, row in s2.iterrows():

        key = (
            row["country_block_key"],
            row["address_block_key"]
        )

        if key != ("", ""):
            address_index.setdefault(key, []).append(
                row["entity_id"]
            )
    for _, row in s1.iterrows():

        name_key = (
            row["country_block_key"],
            row["name_block_key"]
        )

        address_key = (
            row["country_block_key"],
            row["address_block_key"]
        )

        candidates = set()

        candidates.update(
            name_index.get(name_key, [])
        )

        candidates.update(
            address_index.get(address_key, [])
        )

        for source2_id in candidates:

            candidate_pairs.add(
                (
                    row["entity_id"],
                    source2_id
                )
            )

    return pd.DataFrame(
        list(candidate_pairs),
        columns=[
            "source1_id",
            "source2_id"
        ]
    )
```

```python
%%writefile src/business_entity_resolution/features.py

import pandas as pd
from rapidfuzz.fuzz import ratio


def similarity(value1, value2):

    if pd.isna(value1) or pd.isna(value2):
        return 0.0

    value1 = str(value1)
    value2 = str(value2)

    if not value1 or not value2:
        return 0.0

    return ratio(value1, value2) / 100.0


def calculate_pair_features(record1, record2):

    name_score = similarity(
        record1.get("normalized_name", ""),
        record2.get("normalized_name", "")
    )

    address_score = similarity(
        record1.get("normalized_address", ""),
        record2.get("normalized_address", "")
    )

    country_score = similarity(
        record1.get("normalized_country", ""),
        record2.get("normalized_country", "")
    )

    return {
        "name_similarity": name_score,
        "address_similarity": address_score,
        "country_similarity": country_score,
    }


def create_features(source1, source2, candidates):

    s1 = source1.set_index("entity_id")
    s2 = source2.set_index("entity_id")

    rows = []

    for _, candidate in candidates.iterrows():

        id1 = candidate["source1_id"]
        id2 = candidate["source2_id"]

        if id1 not in s1.index or id2 not in s2.index:
            continue

        features = calculate_pair_features(
            s1.loc[id1],
            s2.loc[id2]
        )

        rows.append({
            "source1_id": id1,
            "source2_id": id2,
            **features
        })

    return pd.DataFrame(rows)
```

```python
%%writefile src/business_entity_resolution/model.py

import pandas as pd

from sklearn.linear_model import LogisticRegression


FEATURE_COLUMNS = [
    "name_similarity",
    "address_similarity",
    "country_similarity",
]


def create_weighted_score(features):

    return (
        0.60 * features["name_similarity"]
        + 0.30 * features["address_similarity"]
        + 0.10 * features["country_similarity"]
    )


def train_model(features, labels):

    X = features[FEATURE_COLUMNS]
    y = labels

    model = LogisticRegression(
        max_iter=1000
    )

    model.fit(X, y)

    return model


def predict_scores(model, features):

    X = features[FEATURE_COLUMNS]

    return model.predict_proba(X)[:, 1]
```

```python
%%writefile src/business_entity_resolution/evaluation.py

import pandas as pd
from sklearn.metrics import precision_score, recall_score, fbeta_score


def evaluate_predictions(
    y_true,
    y_pred
):

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f05 = fbeta_score(
        y_true,
        y_pred,
        beta=0.5,
        zero_division=0
    )

    return {
        "precision": precision,
        "recall": recall,
        "f0.5": f05
    }


def evaluate_thresholds(
    scores,
    labels,
    thresholds=None
):

    if thresholds is None:
        thresholds = [
            0.50,
            0.55,
            0.60,
            0.65,
            0.70,
            0.75,
            0.80,
            0.85,
            0.90,
            0.95
        ]

    results = []

    for threshold in thresholds:

        predictions = (
            scores >= threshold
        ).astype(int)

        metrics = evaluate_predictions(
            labels,
            predictions
        )

        results.append({
            "threshold": threshold,
            **metrics
        })

    return pd.DataFrame(results)
```

```python
%%writefile src/business_entity_resolution/inference.py

import pandas as pd


def generate_matches(
    candidates,
    scores,
    threshold=0.80
):

    result = candidates.copy()

    result["match_score"] = scores

    result["is_match"] = (
        result["match_score"] >= threshold
    )

    return result


def build_matching_results(
    source1,
    matched_candidates
):

    matches = (
        matched_candidates[
            matched_candidates["is_match"]
        ]
        .groupby("source1_id")["source2_id"]
        .apply(list)
        .to_dict()
    )

    output = []

    for entity_id in source1["entity_id"]:

        matched_ids = matches.get(
            entity_id,
            []
        )

        output.append({
            "source1_entity_id": entity_id,
            "matched_entity_ids": ",".join(
                matched_ids
            )
        })

    return pd.DataFrame(output)
```

```python
import pandas as pd

source1 = pd.DataFrame({
    "entity_id": ["S1_001", "S1_002"],
    "business_name": ["ABC Technologies Pvt Ltd", "Global Traders"],
    "business_address": ["Hyderabad India", "Chennai India"],
    "country": ["India", "India"]
})

source2 = pd.DataFrame({
    "entity_id": ["S2_001", "S2_002", "S2_003"],
    "business_name": ["ABC Technologies", "Global Trading Company", "Random Business"],
    "business_address": ["Hyderabad India", "Chennai India", "Mumbai India"],
    "country": ["India", "India", "India"]
})

print(source1)
print(source2)
```

```python
from business_entity_resolution.blocking import create_blocking_keys

s1_blocked = create_blocking_keys(source1)
s2_blocked = create_blocking_keys(source2)

print(s1_blocked)
```

```python
print(s2_blocked)
```

```python
from business_entity_resolution.candidate_generation import generate_candidates

candidates = generate_candidates(
    s1_blocked,
    s2_blocked
)

print("Candidate pairs:")
display(candidates)
```

```python
from business_entity_resolution.preprocessing import (
    normalize_name,
    normalize_address,
    normalize_country
)

for df in [s1_blocked, s2_blocked]:
    df["normalized_name"] = df["business_name"].apply(normalize_name)
    df["normalized_address"] = df["business_address"].apply(normalize_address)
    df["normalized_country"] = df["country"].apply(normalize_country)

print("Normalization completed.")
display(s1_blocked)
```

```python
from business_entity_resolution.features import create_features

features = create_features(
    s1_blocked,
    s2_blocked,
    candidates
)

print("Feature generation completed.")
display(features)
```

```python
from business_entity_resolution.model import create_weighted_score

features["weighted_score"] = create_weighted_score(features)

print("Weighted scores:")
display(features)
```

```python
from business_entity_resolution.inference import generate_matches

matches = generate_matches(
    candidates,
    features["weighted_score"],
    threshold=0.80
)

print("Match decisions:")
display(matches)
```

```python
from business_entity_resolution.evaluation import evaluate_predictions

# Expected labels in the SAME order as the matches table
true_labels = [0, 1]

# Predictions produced by our model
predicted_labels = matches["is_match"].astype(int).tolist()

print("True labels:     ", true_labels)
print("Predicted labels:", predicted_labels)

metrics = evaluate_predictions(
    true_labels,
    predicted_labels
)

print("\nEvaluation Results:")
print(metrics)
```

```python
from business_entity_resolution.evaluation import evaluate_thresholds

threshold_results = evaluate_thresholds(
    scores=features["weighted_score"].values,
    labels=true_labels
)

display(threshold_results)
```
