```python
%%writefile src/business_entity_resolution/blocking.py

import re
import pandas as pd


def make_block_key(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = re.sub(r"[^a-z0-9]+", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    if not text:
        return ""
    words = text.split()

    return " ".join(words[:2])


def create_blocking_keys(df):
    result = df.copy()

    result["name_block_key"] = (
        result["business_name"]
        .fillna("")
        .apply(make_block_key)
    )

    result["address_block_key"] = (
        result["business_address"]
        .fillna("")
        .apply(make_block_key)
    )

    result["country_block_key"] = (
        result["country"]
        .fillna("")
        .astype(str)
        .str.lower()
        .str.strip()
    )

    return result
```
