```python
%%writefile src/business_entity_resolution/evaluation.py

import pandas as pd
from sklearn.metrics import precision_score, recall_score, fbeta_score


def evaluate_predictions(
    y_true,
    y_pred
):

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f05 = fbeta_score(
        y_true,
        y_pred,
        beta=0.5,
        zero_division=0
    )

    return {
        "precision": precision,
        "recall": recall,
        "f0.5": f05
    }


def evaluate_thresholds(
    scores,
    labels,
    thresholds=None
):

    if thresholds is None:
        thresholds = [
            0.50,
            0.55,
            0.60,
            0.65,
            0.70,
            0.75,
            0.80,
            0.85,
            0.90,
            0.95
        ]

    results = []

    for threshold in thresholds:

        predictions = (
            scores >= threshold
        ).astype(int)

        metrics = evaluate_predictions(
            labels,
            predictions
        )

        results.append({
            "threshold": threshold,
            **metrics
        })

    return pd.DataFrame(results)
```
